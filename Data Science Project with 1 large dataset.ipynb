```python
import pandas as pd

jobs = pd.read_parquet("C:\\Users\\Lamees\\Downloads\\jobs-00000-of-00001 (1).parquet")
jobs.to_csv("C:\\Users\\Lamees\\Downloads\\jobs-00000-of-00001 (1).parquet", index=False)
jobs.head()
```

```python
#Drop column
fact_job_postings = fact_job_postings.drop(["job_id", "company", "country", "region", "platform", "experience_level", "employment_type", "salary_min_usd", "remote_option", "applicants_estimate", "salary_max_usd"], axis=1)
fact_job_postings.head()
```

```python
#Drop column
global_ai_ml_jobs_2025 = global_ai_ml_jobs_2025.drop(["Experience Level", "Employment Type", "Company Size", "Work Setting", "Country", "Salary (USD)"], axis=1)
global_ai_ml_jobs_2025.head()
```

```python
#Drop column
data_jobs = data_jobs.drop(["job_title_short", "job_location", "job_via", "job_schedule_type", "job_work_from_home", "search_location", "job_no_degree_mention", "job_health_insurance", "job_country", "salary_rate", "salary_year_avg", "salary_hour_avg", "salary_hour_avg", "company_name"], axis=1)

#Join columns with space
data_jobs["Skills"] = data_jobs["job_skills"] + " " + data_jobs["job_type_skills"]

data_jobs.head()
```

```python
#Drop skill columns as they are all combined in the "Skills" column
data_jobs = data_jobs.drop(["job_skills", "job_type_skills"], axis=1)
data_jobs.head()
```

```python
#Drop columns that are not needed
jobs = jobs.drop(["company_name", "company_domain", "company_industry", "ats", "ats_token", "job_id", "department", "location_raw", "country", "employment_type", "salary_min", "salary_max", "salary_currency", "scraped_at", "url"], axis=1)
jobs = jobs.drop(["city", "is_remote", "salary_period", "salary_from_text"], axis=1)
jobs.head()
```

```python
data_jobs_clean = data_jobs.dropna()
data_jobs_clean.head(10)
```

```python
#Checking how many empty rows exist in the jobs dataset
import numpy as np
empty_count = jobs["skills"].apply(lambda x: isinstance(x, np.ndarray) and x.size == 0).sum()
print("Empty array rows:", empty_count)
```

```python
#Drop rows with empty arrays in skills in the dataset jobs
jobs_clean = jobs[jobs["skills"].apply(lambda x: not (isinstance(x, np.ndarray) and x.size == 0))]

jobs_clean.head(10)
```

```python
#Rename column names to ensure standardisation
fact_job_postings = fact_job_postings.rename(columns={"job_title": "job_title", "posting_date": "posting_date"})
global_ai_ml_jobs_2025 = global_ai_ml_jobs_2025.rename(columns={"Job Title": "job_title", "Skills": "skills", "Year": "posting_date"})
data_jobs_clean = data_jobs_clean.rename(columns={"job_title": "job_title", "Skills": "skills", "job_posted_date" : "posting_date"})
jobs_clean = jobs_clean.rename(columns={"title": "job_title", "skills": "skills", "posted_at": "posting_date"})
```
